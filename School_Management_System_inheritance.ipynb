```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show_details(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")


class Student(Person):
    def __init__(self, name, age, student_id, marks):
        super().__init__(name, age)
        self.student_id = student_id
        self.marks = marks              # dict: {"English": 85, "Math": 90, ...}

    def show_details(self):
        super().show_details()          # reuse parent's name/age output
        print(f"Student ID: {self.student_id}")

    def calculate_total(self):
        return sum(self.marks.values())

    def calculate_percentage(self):
        return (self.calculate_total() / (len(self.marks) * 100)) * 100

    def calculate_grade(self):
        # Base rule — regular grading (overridden by ScholarshipStudent)
        pct = self.calculate_percentage()
        if pct >= 90:
            return "A+"
        elif pct >= 80:
            return "A"
        elif pct >= 70:
            return "B"
        elif pct >= 60:
            return "C"
        elif pct >= 50:
            return "D"
        else:
            return "Fail"

    def display_result(self):
        print()
        print(f"===== Result: {self.name} (ID: {self.student_id}) =====")
        print("Marks:")
        for subject, mark in self.marks.items():
            print(f"  {subject}: {mark}")
        print(f"Total: {self.calculate_total()} / {len(self.marks) * 100}")
        print(f"Percentage: {self.calculate_percentage():.2f}%")
        print(f"Grade: {self.calculate_grade()}")


class ScholarshipStudent(Student):
    def calculate_grade(self):
        # Scholarship grading is stricter — higher thresholds
        pct = self.calculate_percentage()
        if pct >= 95:
            return "A+ (Scholarship)"
        elif pct >= 85:
            return "A (Scholarship)"
        elif pct >= 75:
            return "B (Scholarship)"
        elif pct >= 65:
            return "C (Scholarship)"
        elif pct >= 55:
            return "D (Scholarship)"
        else:
            return "Fail (Scholarship risk)"


class Teacher(Person):
    def __init__(self, name, age, employee_id, subject):
        super().__init__(name, age)
        self.employee_id = employee_id
        self.subject = subject

    def show_details(self):
        super().show_details()
        print(f"Employee ID: {self.employee_id}")
        print(f"Subject: {self.subject}")


# ---- Storage ----
students = {}
teachers = {}


def input_marks():
    marks = {}
    print("Enter marks for 5 subjects (0-100):")
    for subject in ["English", "Math", "Computer", "Science", "Urdu"]:
        while True:
            try:
                mark = float(input(f"  {subject}: "))
                if 0 <= mark <= 100:
                    marks[subject] = mark
                    break
                else:
                    print("  Marks must be between 0 and 100.")
            except ValueError:
                print("  Invalid number, try again.")
    return marks


def add_student():
    print()
    print("===== ADD STUDENT =====")
    print("1. Regular Student")
    print("2. Scholarship Student")
    choice = input("Enter choice (1-2): ").strip()

    if choice not in ("1", "2"):
        print("Invalid choice.")
        return

    name = input("Enter name: ")
    try:
        age = int(input("Enter age: "))
    except ValueError:
        print("Invalid age.")
        return

    student_id = input("Enter student ID: ").strip()
    if student_id in students:
        print(f"Student ID {student_id} already exists.")
        return

    marks = input_marks()

    if choice == "1":
        students[student_id] = Student(name, age, student_id, marks)
        print(f"Regular student {name} added.")
    else:
        students[student_id] = ScholarshipStudent(name, age, student_id, marks)
        print(f"Scholarship student {name} added.")


def add_teacher():
    print()
    print("===== ADD TEACHER =====")
    name = input("Enter name: ")
    try:
        age = int(input("Enter age: "))
    except ValueError:
        print("Invalid age.")
        return

    emp_id = input("Enter employee ID: ").strip()
    if emp_id in teachers:
        print(f"Employee ID {emp_id} already exists.")
        return

    subject = input("Enter subject: ")
    teachers[emp_id] = Teacher(name, age, emp_id, subject)
    print(f"Teacher {name} added.")


def view_student_result():
    student_id = input("Enter student ID: ").strip()
    if student_id not in students:
        print(f"Student ID {student_id} not found.")
        return
    students[student_id].display_result()


def check_subject_marks():
    student_id = input("Enter student ID: ").strip()
    if student_id not in students:
        print(f"Student ID {student_id} not found.")
        return
    subject = input("Enter subject name: ").strip()
    student = students[student_id]

    if subject in student.marks:
        print(f"{student.name}'s {subject} marks: {student.marks[subject]}")
    else:
        print(f"Subject '{subject}' not found.")
        print(f"Available subjects: {', '.join(student.marks.keys())}")


def list_all():
    print()
    print("===== ALL STUDENTS =====")
    if not students:
        print("No students registered.")
    else:
        for sid, s in students.items():
            print(f"  {sid}: {s.name} ({type(s).__name__})")

    print()
    print("===== ALL TEACHERS =====")
    if not teachers:
        print("No teachers registered.")
    else:
        for tid, t in teachers.items():
            print(f"  {tid}: {t.name} — {t.subject}")


def main():
    print("=========================================")
    print("      SCHOOL MANAGEMENT SYSTEM           ")
    print("=========================================")

    while True:
        print()
        print("===== MENU =====")
        print("1. Add Student")
        print("2. Add Teacher")
        print("3. View Student Result")
        print("4. Check Subject Marks")
        print("5. List All")
        print("6. Exit")

        choice = input("Enter choice (1-6): ").strip()

        if choice == "1":
            add_student()
        elif choice == "2":
            add_teacher()
        elif choice == "3":
            view_student_result()
        elif choice == "4":
            check_subject_marks()
        elif choice == "5":
            list_all()
        elif choice == "6":
            print("Goodbye!")
            break
        else:
            print("Invalid choice. Please enter 1-6.")


main()
```
